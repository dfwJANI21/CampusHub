```
import time

# ---- Song info -------------------------------------------------------
SONG_TITLE = "I Think They Call This Love (Cover)"
YOUTUBE_URL = "https://youtu.be/CnEqrgMlWLQ?si=PRXAGg14KAJpFWry"

lyrics = [
    (130, "Love what could this be between you and me "),
    (134, "oh oh I dream of is yours all I long for is"),
    (138, "touch and darling something tells me tells me its"),
    (142, "enough you can say that Im fool"),
]

START_FROM = 130


def type_lyric(text: str, speed: float) -> None:
    """Print one line character by character at the given per-char delay."""
    for char in text:
        print(char, end="", flush=True)
        time.sleep(speed)
    print()


def run(lyrics, start_from: float) -> None:
    print(f"Song: {SONG_TITLE}")
    print(f"YouTube: {YOUTUBE_URL}\n")
    print(f"Starting from {int(start_from // 60)}:{int(start_from % 60):02d}...\n")

    start_time = time.perf_counter() - start_from
    n = len(lyrics)

    for i, (timestamp, lyric) in enumerate(lyrics):
        if timestamp < start_from:
            continue  # skip lines before our starting point

        target_time = start_time + timestamp
        while time.perf_counter() < target_time:
            time.sleep(0.005)

        # Calculate how much time is available before next lyric
        if i < n - 1:
            next_timestamp = lyrics[i + 1][0]
            available_time = next_timestamp - timestamp

            # Calculate typing speed automatically (guarded against blank lines)
            typing_speed = available_time / len(lyric) if lyric else 0.15
        else:
            # Last line
            typing_speed = 0.15

        type_lyric(lyric, typing_speed)


if __name__ == "__main__":
    run(lyrics, START_FROM)
```

### Overwriting vs Creating Files
Opening a file with `'w'` (write mode) will either overwrite an existing file or create it if it doesn't exist.

```
# 1. Creating a brand new file
new_file_path = '/content/brand_new_file.txt'

with open(new_file_path, 'w') as f:
    f.write("I just created this file from scratch!\n")

print(f"File created at {new_file_path}")

# 2. Reading it back to confirm
with open(new_file_path, 'r') as f:
    print("New file content:", f.read())
```

```
secret = 21
guess = 0
guess_count = 0
guess_limit = 10
guess_over = False

while guess != secret and not guess_over:
    try:
        if guess_count < guess_limit:
            guess_str = input("Enter the secret number: ")
            guess = int(guess_str)
            guess_count += 1
            print("total try: " + str(guess_count))

            if guess > secret:
                print("less")
            elif guess < secret:
                print(
                "more")
            else:
                print("You guessed it correctly!")
        else:
            guess_over = True
            print("You ran out of guesses!")
            break

    except ValueError:
        print("Invalid input. Please enter a valid number.")

```
